Partie 1 : Listes Python vs tableaux NumPy On dispose des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8  
1) Créer une liste Python stockant ces températures;
 2) Créer un tableau NumPy équivalent stockant ces températures ;
 3) Effectuer une multiplication par 2 sur ces deux collections;
  4) Quelle différence observe-t-on ?

```python
!pip install numpy
```

```python
import numpy as np
```

```python
# creer une liste python stokant les temperature
listes_simple=[24.5, 25.1, 26.3, 27.0 ,25.8]
print (listes_simple)
```

```python
#Créer un tableau NumPy équivalent stockant ces températures
listes_np=np.array([24.5, 25.1, 26.3, 27.0 ,25.8])
print (listes_np)
```

```python
## Effectuer une multiplication par 2 sur ces deux collections;
manup=listes_np*2
print (manup)
```

```python
# Effectuer une multiplication par 2 sur ces deux collections;
manup1=listes_simple*2
print (manup1)
```

Quelle différence observe-t-on ?

liste_simple * 2 → duplique la liste
 (tu obtiens 10 éléments : la liste répétée deux fois).
liste_np * 2 → multiplie chaque température par 2
(tu obtiens toujours 5 éléments, mais chacun doublé).

C'est exactement la démonstration qu'un tableau NumPy fait du vrai calcul, une liste Python non.



Partie 2 – Création des données
 1) Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro.
 2) Créer un tableau contenant 10 valeurs égales à 1.
 3) Créer un tableau de 10 valeurs initialisées à 25.
  4) Créer une matrice identité de taille 5 × 5.
  5) Créer les 24 heures (de 0 à 23) de mesure d’une journée
  6) Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C
  7) Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles

```python
#Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro.
temperature_zero=np.zeros(10)
print (temperature_zero)
```

```python
#Créer un tableau contenant 10 valeurs égales à 1.
temperature_un=np.ones(10)
print (temperature_un)
```

```python
#Créer un tableau de 10 valeurs initialisées à 25.
temperature_25=np.full(10,25)
print (temperature_25)
```

```python
#Créer une matrice identité de taille 5 × 5.
matrice_identite=np.eye(5)
print (matrice_identite)
```

```python
#Créer les 24 heures (de 0 à 23) de mesure d’une journée
heure=np.arange(24)
print (heure)
```

```python
#Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C
temperature_20_30=np.linspace(20,30,10)
print (temperature_20_30)
```

```python
#Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductible
np.random.seed(7)
temperature_aleatoire=np.random.randint(15,39,100)
print (temperature_aleatoire)
```
